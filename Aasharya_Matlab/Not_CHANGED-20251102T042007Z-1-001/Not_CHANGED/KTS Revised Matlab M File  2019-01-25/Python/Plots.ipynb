```python
import scipy.io as sio
import numpy as np
import matplotlib.pyplot as plt

# Load MATLAB files
data1 = sio.loadmat("firstoutput.mat")
data2 = sio.loadmat("dustoutput13.mat")
data3 = sio. loadmat("dustoutput13cold.mat")
# Extract variables
x = data1["X"].flatten() / data1["DLe"].flatten()
```

# Ion density comparison

```python
plt.figure(figsize=(8,6))
plt.plot(x, data1["ni"].flatten(), "r-", linewidth=1.8, label="Non-Dust Plasma")
plt.plot(x, data2["ni"].flatten(), "b-", linewidth=1.8, label="Dusty Plasma")

plt.xlabel(r"$X/D_{Le}$", fontsize=14)
plt.ylabel(r"Ion Density ($n_i$)", fontsize=14)
plt.title("Ion Density Comparison", fontsize=16)
plt.xlim(0,10)
plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Electron density comparison

```python
plt.figure(figsize=(8,6))
plt.plot(x, data1["ne"].flatten(), "r-", linewidth=1.8, label="Non-Dust Plasma")
plt.plot(x, data2["ne"].flatten(), "b-", linewidth=1.8, label="Dusty Plasma")

plt.xlabel(r"$X/D_{Le}$", fontsize=14)
plt.ylabel(r"Electron Density ($n_e$)", fontsize=14)
plt.title("Electron Density Comparison", fontsize=16)
plt.xlim(0,10)
plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Potential Evolutiom Comparison

```python
plt.figure(figsize=(8,6))
plt.plot(x, data1["phi"].flatten(), "r-", linewidth=1.8, label="Non-Dust Plasma")
plt.plot(x, data2["phi"].flatten(), "b-", linewidth=1.8, label="Dusty Plasma")

plt.xlabel(r"$X/D_{Le}$", fontsize=14)
plt.ylabel(r"$\Phi$", fontsize=14)
plt.title("Potential Evolution Comparison", fontsize=16)
plt.legend(fontsize=12)
plt.xlim(0,10)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Normalized Dust Charge

```python
e = 1.6022e-19
normalized_charge = data2["Qdn"].flatten() / e

plt.figure(figsize=(8,6))
plt.plot(x, normalized_charge, "m-", linewidth=1.8)

plt.xlabel(r"$X/D_{Le}$", fontsize=14)
plt.ylabel("Normalized Dust Charge ($Q_d$)", fontsize=14)
plt.title("Normalized Dust Charge", fontsize=16)
plt.xlim(0,10)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

# Space Charge Density comparison

```python
plt.figure(figsize=(8,6))
plt.plot(x, data1["rho"].flatten(), "r-", linewidth=1.8, label="Non-Dust Plasma")
plt.plot(x, data2["rho"].flatten(), "b-", linewidth=1.8, label="Dusty Plasma")

plt.xlabel(r"$X/D_{Le}$", fontsize=14)
plt.ylabel(r"Space Charge Density ($\rho$)", fontsize=14)
plt.title("Space Charge Density Comparison", fontsize=16)
plt.legend(fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

 # Temperature comparision Cold vs Hot plasma
 

```python
fig, axes = plt.subplots(1, 2, figsize=(14,6), sharex=True)

# --- Left subplot: Dust potential ---
axes[0].plot(x, data2["phid"].flatten(), "r-", linewidth=1.8, label=r"$\sigma > 0$")
axes[0].plot(x, data3["phid"].flatten(), "b-", linewidth=1.8, label=r"$\sigma \simeq 0$")
axes[0].set_xlabel(r"$X/D_{Le}$", fontsize=14)
axes[0].set_ylabel(r"$\phi_d$", fontsize=14)
axes[0].set_title("Dust Potential vs Normalized Distance", fontsize=15)
axes[0].legend(fontsize=12)
axes[0].set_xlim(0,10)
axes[0].grid(True, linestyle="--", alpha=0.6)

# --- Right subplot: Plasma potential ---
axes[1].plot(x, data1["phi"].flatten(), "--", linewidth=1.8, label=r"$non-dust \sigma > 0$")
axes[1].plot(x, data2["phi"].flatten(), "r-", linewidth=1.8, label=r"$\sigma > 0$")
axes[1].plot(x, data3["phi"].flatten(), "b-", linewidth=1.8, label=r"$\sigma \simeq 0$")
axes[1].set_xlabel(r"$X/D_{Le}$", fontsize=14)
axes[1].set_ylabel(r"$\phi$", fontsize=14)
axes[1].set_title("Plasma Potential vs Normalized Distance", fontsize=15)
axes[1].legend(fontsize=12)
axes[1].set_xlim(0,10)
axes[1].grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
